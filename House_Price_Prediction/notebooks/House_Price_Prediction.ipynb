## 9.1 Separate Features & Target

**Purpose:** Separate the input features (`X`) from the target variable (`y`). The model learns from `X` and predicts `Price`.

```python
# Features
X = df_clean.drop(columns=["Price"])

# Target
y = df_clean["Price"]

print("Feature Matrix (X):", X.shape)
print("Target Vector (y):", y.shape)
```

## 9.2 One-Hot Encoding

**Purpose:** Convert categorical variables into numerical binary features using One-Hot Encoding.

### Why do we use `drop_first=True`?

One category from each categorical feature is automatically removed as the **reference category**. This prevents perfect multicollinearity (Dummy Variable Trap).

### Reference Categories

- **Location:** Downtown
- **Condition:** Excellent
- **Garage:** No

```python
X_encoded = pd.get_dummies(
    X,
    columns=["Location", "Condition", "Garage"],
    drop_first=True,
    dtype=int
)

print("Original Shape :", X.shape)
print("Encoded Shape  :", X_encoded.shape)
```

## 9.3 Verify Encoded Features

**Purpose:** Confirm that the categorical variables were converted into the correct binary feature columns.

```python
X_encoded.columns
```

## 9.4 Train/Test Split

**Purpose:** Divide the dataset into training and testing sets for unbiased model evaluation.

### Split Strategy

- **Training:** 80% (1,600 houses)
- **Testing:** 20% (400 houses)

`random_state=42` ensures reproducibility.

```python
# Features and Target
X = df_clean.drop(columns=["Price"])
y = df_clean["Price"]

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

## 9.5 Feature Scaling

**Purpose:** Standardize only the numerical features using statistics learned from the training set.

### Why fit only on training data?

To prevent **data leakage**. The scaler learns the mean and standard deviation from the training set only, then applies the same transformation to the test set.

```python
from sklearn.preprocessing import StandardScaler

numerical_cols = [
    "Area",
    "Bedrooms",
    "Bathrooms",
    "Floors",
    "YearBuilt"
]

scaler = StandardScaler()

X_train[numerical_cols] = scaler.fit_transform(X_train[numerical_cols])
X_test[numerical_cols] = scaler.transform(X_test[numerical_cols])

print("Feature scaling completed.")
```

## 9.6 Verify the Final ML Dataset

**Purpose:** Confirm that the processed datasets are fully prepared for machine learning model development.

```python
print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)
print("Training Target   :", y_train.shape)
print("Testing Target    :", y_test.shape)

X_train.head()
```

The testing dataset is completely isolated from the learning process. It is only used after training to measure how well the model generalizes to unseen houses.

## 21.2 Create New Features

**Purpose:** Generate additional informative features from existing variables.

Feature engineering often improves model performance by exposing relationships that were not explicitly present in the original dataset.

```python
df_improved = df_clean.copy()

# House age
df_improved["HouseAge"] = 2026 - df_improved["YearBuilt"]

# Total rooms
df_improved["TotalRooms"] = (
    df_improved["Bedrooms"] +
    df_improved["Bathrooms"]
)

df_improved.head()
```

## 21.3 Remove Weak Features

**Purpose:** Remove features that contribute little information.

Based on the previous statistical analysis, `Id` is not a meaningful predictor and should not be used for machine learning.

Only meaningful property characteristics should remain.

```python
X_new = df_improved.drop(columns=["Price"])
y_new = df_improved["Price"]

print(X_new.shape)
```

## 21.4 Retrain the Improved Model

**Purpose:** Train the improved Linear Regression model using the newly engineered features.

```python
X_train_new, X_test_new, y_train_new, y_test_new = train_test_split(
    X_new,
    y_new,
    test_size=0.20,
    random_state=42
)

improved_model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])

improved_model.fit(X_train_new, y_train_new)

y_pred_new = improved_model.predict(X_test_new)
```

## 21.5 Compare Before vs After

**Purpose:** Compare the original baseline with the improved model.

```python
new_mae = mean_absolute_error(y_test_new, y_pred_new)
new_rmse = np.sqrt(mean_squared_error(y_test_new, y_pred_new))
new_r2 = r2_score(y_test_new, y_pred_new)

comparison = pd.DataFrame({
    "Metric": ["MAE", "RMSE", "R²"],
    "Before": [243241.98, 279859.73, -0.0067],
    "After": [new_mae, new_rmse, new_r2]
})

comparison
```

- Retrained the model
- Compared original and improved performance

### Engineering Interpretation

Feature engineering is often more valuable than simply trying additional algorithms. Better features can improve model performance because they provide the model with more meaningful information about the underlying problem.
